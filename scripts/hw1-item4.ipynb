```python
import pandas as pd
```

```python
df = pd.read_excel("DragAndAngle.xlsx")
```

```python
df = df.rename(columns= {"Column1":"Angle of Attack","Column2":"Drag Coefficient"})
```

```python
df["Angle of Attack"] = pd.to_numeric(df["Angle of Attack"], errors= "coerce")
df["Drag Coefficient"] = pd.to_numeric(df["Drag Coefficient"],errors= "coerce")
```

```python
df.dropna(how="all")
```

```python
df.isnull()
```

```python
df.isnull().sum()
```

```python
df = df.dropna().reset_index(drop=True)
```

```python
print(df)
```

```python
df = df[(df["Angle of Attack"]>0) & (df["Drag Coefficient"]>0) & (df["Drag Coefficient"]<1)]
```

```python
print(df)
```
